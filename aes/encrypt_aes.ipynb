# AES-256-GCM Encryption

AES-GCM is an *authenticated* cipher: it produces the encrypted message
(**ciphertext**) plus a **tag** that proves nobody modified it.

Every call creates a **fresh random 12-byte nonce**. Reusing a nonce with
the same key breaks GCM's security completely, so we generate a new one
each time with `get_random_bytes(12)`.

```python
from typing import Tuple

from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes


def encrypt(key: bytes, plaintext: bytes, aad: bytes = b"") -> Tuple[bytes, bytes, bytes]:
    """Encrypt plaintext with AES-GCM (authenticated encryption).

    Returns (nonce, ciphertext, tag). The optional `aad` is authenticated
    but left unencrypted.
    """
    cipher = AES.new(key, AES.MODE_GCM, nonce=get_random_bytes(12))
    cipher.update(aad)  # authenticate the additional data
    ciphertext, tag = cipher.encrypt_and_digest(plaintext)
    return cipher.nonce, ciphertext, tag
```

## Try it

Encrypt a message and look at the nonce, ciphertext and tag (shown in hex).
Run the cell twice — the ciphertext changes every time because the nonce
is different on every call.

```python
from generate_aes import generate_key

key = generate_key()  # 256-bit random key
nonce, ciphertext, tag = encrypt(key, b"Hello, AES!")

print("Nonce      :", nonce.hex())
print("Ciphertext :", ciphertext.hex())
print("Tag        :", tag.hex())
```
