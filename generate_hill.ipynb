# Hill Cipher Key Generation

The Hill Cipher requires an $n \times n$ key matrix $K$ with entries in $\mathbb{Z}_{26}$.
For the cipher to be decipherable, $K$ must be invertible modulo 26:

$$\gcd(\det(K) \pmod{26}, 26) = 1$$

Because $26 = 2 \times 13$, this means $\det(K)$ must not be divisible by 2 or 13.

```python
import math
import secrets
from typing import List
import sympy


def matrix_mod_inverse(matrix: List[List[int]], modulus: int = 26) -> List[List[int]]:
    m = sympy.Matrix(matrix)
    if m.rows != m.cols:
        raise ValueError(f"Key matrix must be square, got {m.rows}x{m.cols}")
    det = int(m.det()) % modulus
    if math.gcd(det, modulus) != 1:
        raise ValueError(f"Matrix not invertible mod {modulus} (gcd({det}, {modulus}) != 1)")
    inv = m.inv_mod(modulus)
    return [[int(inv[r, c]) for c in range(inv.cols)] for r in range(inv.rows)]


def generate_key(n: int = 2) -> List[List[int]]:
    if n < 2:
        raise ValueError("Key matrix dimension n must be at least 2")
    while True:
        candidate = [[secrets.randbelow(26) for _ in range(n)] for _ in range(n)]
        det = int(sympy.Matrix(candidate).det()) % 26
        if math.gcd(det, 26) == 1:
            return candidate
```

## Generate and Inspect Key Matrix

Generate a $2 \times 2$ invertible key matrix and verify its modular inverse.

```python
key = generate_key(2)
inv_key = matrix_mod_inverse(key, 26)
det = int(sympy.Matrix(key).det()) % 26

print("Key Matrix (K):       ", key)
print("Determinant mod 26:   ", det)
print("Inverse Matrix (K^-1):", inv_key)
```
