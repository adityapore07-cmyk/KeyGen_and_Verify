```python
from generate_rsa import generate_new_rsa_pair

def encrypt(text):
    key = generate_new_rsa_pair(1024)
    t = int.from_bytes(text, byteorder='big') 
    n = key[0][0]
    e = key[0][1]
    d = key[1][1]
    res = (t ** int(e)) % int(n)
    return res, n, d
```

```python
encrypted, n, d = encrypt(b"Hello World")
# print(msg, "\n", n, "\n", d, "\n", l)
encrypted
```
