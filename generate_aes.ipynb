# AES Key Generation

AES-GCM needs a symmetric key of **16, 24 or 32 bytes** (128, 192 or 256 bits).
We generate it with `Crypto.Random.get_random_bytes`, PyCryptodome's
cryptographically secure random generator.

Never hard-code a key and never reuse one — always generate a fresh key.

```python
from Crypto.Random import get_random_bytes


def generate_key(bits: int = 256) -> bytes:
    """Generate a random AES key of 128, 192 or 256 bits."""
    if bits not in (128, 192, 256):
        raise ValueError("Key size must be 128, 192 or 256 bits")
    return get_random_bytes(bits // 8)
```

Generate a **256-bit (32-byte)** key.

We only print the *length* — a secret key should never be printed or stored
next to the ciphertext it protects.

```python
key = generate_key()
print("Key length (bytes):", len(key))
```

Try `generate_key(128)` for a shorter key, or `generate_key(512)` to see
the size check raise a `ValueError`.
