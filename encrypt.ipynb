```python
from generate import generate_new_pair

def encrypt(text):
    res = ""
    key = generate_new_pair(1024)
    n = key[0][0]
    e = key[0][1]
    print(e)
    # d = key[1][1]
    for t in text:
        o = ord(t)
        ch = str((int(o) ^ int(e)) % int(n))
        # l = len(str(n))
        res += ch + ("0" * (5 - len(ch)))
    return res
```
